Example 1: Sum of Even Numbers

```python
def sum_even_numbers(n):   # def sum_even_numbers(n):
  result = 0

  for i in range(1, n + 1):
    if i % 2 == 0:
      result += i

    return result


print("sum_even_numbers:", sum_even_numbers(20))

                          # def → used to create a function.
                            # sum_even_numbers → function name.
                            # n → parameter/input.
                            #: → starts the function body.
```

Example 2: Factorial with While Loop

```python
def factorial(n):
  result = 1
  while n > 0:
    result *= n
    n -= 1
  return result

print(factorial(5)) # Example call
```

reverse a string

```python
def reverse_string(s):
  return s[::-1]
print(reverse_string("naresh"))
```

using loop return a reverse string

```python
def reverse_string(s):
  reverse = ""

  for ch in s:
    reverse = ch + reverse
  return reverse

print(reverse_string("naresh"))
```

calculate bmi

```python
def cal_bmi(weight,height):
  bmi = weight / (height ** 2)

  return "underweight" if bmi < 18.5 else "normal" if 18.5 <= bmi > 25 else "overweight"

print(cal_bmi(50,1.63))
```

count vowels in string

```python
def count_vowels(s):
  count = 0

  for char in s:
    if char.lower() in "a,e,i,o,u":
      count = count + 1
  return count

print(count_vowels("naresh"))
```

sum of digits

```python
def sum_of_digits(number):
  return sum(int(digit) for digit in str(abs(number)))

print(sum_of_digits(143))
```

```python
def sum_of_digit(num):
  number = abs(num)
  total = 0

  while num > 0:
    digit = num % 10
    total += digit
    num //= 10

    return total

print(sum_of_digit(78))
```

pyramid

```python
def print_pyramid(rows):

  for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i -1))

print("pyramid pattern with 5 rows:")
print_pyramid(5)
```
