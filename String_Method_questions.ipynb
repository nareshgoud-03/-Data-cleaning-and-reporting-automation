Q25. Create a menu-driven program where user can select string operations like reverse, capitalize, find length, etc.

```python
text = input("enter a string")

print("1.reverse a string")
print("2.capitalize a string")
print("3.length of a string")

choice = int(input("enter your choice"))

if choice == 1:
  print(text[::-1])

elif choice == 2:
  print(text.capitalize())

elif choice == 3:
  print(len(text))

else:
  print("invalid")
```
